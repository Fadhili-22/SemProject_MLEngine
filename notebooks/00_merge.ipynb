```python
#necessary imports
import pandas as pd
```

```python
#loading the five original files
demographic = pd.read_csv('../data/demographic.csv')
phq9 = pd.read_csv('../data/phq9.csv')
gad7 = pd.read_csv('../data/gad7.csv')
isi = pd.read_csv('../data/isi.csv')
pss = pd.read_csv('../data/pss.csv')
```

```python
#renaming the columns so each one is clear
demographic.columns = demographic.columns.str.replace('edu', 'education').str.replace('smoke', 'smoking').str.replace('drink', 'drinking')
phq9.columns = phq9.columns.str.replace('score', 'phq_total').str.replace('question', 'phq_q').str.replace('time', 'phq_t')
gad7.columns = gad7.columns.str.replace('score', 'gad_total').str.replace('question', 'gad_q').str.replace('time', 'gad_t')
isi.columns = isi.columns.str.replace('score', 'isi_total').str.replace('question', 'isi_q').str.replace('time', 'isi_t')
pss.columns = pss.columns.str.replace('score', 'pss_total').str.replace('question', 'pss_q').str.replace('time', 'pss_t')
```

```python
#joining the five files on the student id
df = demographic.merge(phq9, on='export_id').merge(gad7, on='export_id').merge(isi, on='export_id').merge(pss, on='export_id')
df.shape
```

```python
#saving the combined dataset for notebook 01
df.to_csv('../data/zenodo.csv', index=False)
```
