### 2.6 Data Cleaning

The raw data cannot be used for modelling yet. This section cleans it one step at a time and prints the number of rows and columns after each filter. The steps are:

1. Check the age range.
2. Work out how many seconds each student spent per question.
3. Remove students who scored 0 on the PHQ-9, GAD-7 and ISI.
4. Keep students aged 16 to 35.
5. Remove students who answered too fast to have read the questions.
6. Save the cleaned data.

#### Step 1: Check the Age Range

The summary statistics showed ages from 1 to 99. The next cell shows the full age summary again and the following cell counts how many students are below 16 or above 35.

```python
#checking the age range
df['age'].describe()
```

```python
#students below 16 and above 35
print((df['age'] < 16).sum(), (df['age'] > 35).sum())
```

#### Observation

There are 22 students younger than 16 and 34 students older than 35. Raw ages include 1, 5 and 10 on one side and 80, 81 and 99 on the other, which cannot be university students. These rows are removed in Step 4.

#### Step 2: Average Time per Question

Every question has a time column (for example `phq_t1`) that holds the seconds the student took to answer it. There are 37 of them (9 PHQ-9, 7 GAD-7, 7 ISI and 14 PSS). The new column `avg_time_per_item` is the mean of these 37 columns. The total columns (such as `phq_total`) also contain "_t" in their name, so they are left out.

```python
#average time per question = mean of the 37 time columns (the total columns are not time columns)
time_cols = [c for c in df.columns if '_t' in c and 'total' not in c]
print(len(time_cols))
df['avg_time_per_item'] = df[time_cols].mean(axis=1).round(2)
```

```python
#summary of the average time per question
df['avg_time_per_item'].describe()
```

#### Observation

The median student spends 3.8 seconds per question, but the fastest student averages 0.46 seconds and the slowest 377.8 seconds. Under half a second is too short to read a question, and several minutes per question suggests the student left the page open. The very fast students are removed in Step 5.

#### Step 3: Remove Students Who Scored 0 on All Three Questionnaires

Some students have a total of 0 on the PHQ-9, GAD-7 and ISI. These students may have clicked "never" on every question without reading it. Some of them may truly have no symptoms, so this is a choice that has a cost: the cleaned data has fewer "okay" students than the real population.

```python
#dropping students with phq, gad and isi totals all equal to 0
df = df[~((df['phq_total'] == 0) & (df['gad_total'] == 0) & (df['isi_total'] == 0))]
print(df.shape)
```

#### Observation

This step removes 4,553 students (about 19%) and leaves 19,739 rows. It is the largest filter, so the share of level 0 (okay) students will be lower than in the raw data. We keep it because answering 0 to all 23 questions across three questionnaires is a strong sign of careless answers, but the app's real users may have more healthy students than our data.

#### Step 4: Keep Ages 16 to 35

The app is meant for university students, so only students aged 16 to 35 are kept. Younger ages and ages above 35 are outside the student range or are probably typing mistakes.

```python
#keeping students aged 16 to 35
df = df[(df['age'] >= 16) & (df['age'] <= 35)]
print(df.shape)
```

#### Observation

Only 40 more students are removed, leaving 19,699 rows. Of the 56 students outside the age range, 16 had already been removed in Step 3, so the age filter has a small effect.

#### Step 5: Remove Students Who Answered Too Fast

Students who spent less than 1 second per question on average are removed. At that speed the questions could not have been read, so the answers are probably careless and would add noise to the models.

```python
#dropping students who spent less than 1 second per question
df = df[df['avg_time_per_item'] >= 1]
print(df.shape)
```

#### Observation

This removes 448 students and leaves 19,251 rows. After this step the fastest student averages exactly 1 second per question.

#### Step 6: Save the Cleaned Data

The cleaned data is saved to `data/zenodo_clean.csv`. The next notebooks read this file, not the raw one, so the raw file `zenodo.csv` is never changed.

```python
#saving the cleaned data
df.to_csv('../data/zenodo_clean.csv', index=False)
```

```python
#checking the saved file has the expected number of rows
print(pd.read_csv('../data/zenodo_clean.csv').shape)
print(len(df) == 19251)
```

#### Observation

The saved file has exactly 19,251 rows, as expected. It has 85 columns, which are the 84 original columns plus `avg_time_per_item`.
